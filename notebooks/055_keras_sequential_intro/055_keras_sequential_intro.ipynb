# Keras Sequential — Fashion-MNIST exercise

Following along with the Keras getting-started page. Goal: get a Sequential model end-to-end with `tf.keras` so I have the API in my head. Not aiming for a competitive accuracy here — just want to see the pieces fit together (`compile`, `fit`, `evaluate`).

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras import layers, models

tf.random.set_seed(0)
np.random.seed(0)
print(tf.__version__)
```

```python
(x_tr, y_tr), (x_te, y_te) = tf.keras.datasets.fashion_mnist.load_data()
x_tr = x_tr.astype("float32") / 255.0
x_te = x_te.astype("float32") / 255.0
print(x_tr.shape, x_te.shape)
```

```python
model = models.Sequential([
    layers.Flatten(input_shape=(28, 28)),
    layers.Dense(128, activation="relu"),
    layers.Dense(10),  # logits, since we use from_logits=True below
])
model.compile(optimizer="adam",
              loss=tf.keras.losses.SparseCategoricalCrossentropy(from_logits=True),
              metrics=["accuracy"])
model.summary()
```

```python
# 3 epochs is enough to see the curve do something — not enough to be competitive
history = model.fit(x_tr, y_tr, epochs=3, batch_size=128, validation_split=0.1, verbose=2)
```

```python
test_loss, test_acc = model.evaluate(x_te, y_te, verbose=0)
print(f"test accuracy: {test_acc:.4f}")
```

Quick tutorial run, not a benchmark. The dense-only model treats each pixel as independent, which is obviously the wrong inductive bias for images. A CNN would do a lot better here for not much more code — that's notebook 53.
